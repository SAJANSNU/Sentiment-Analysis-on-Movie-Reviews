```python
# Step 1: Imports
from transformers import BertForSequenceClassification, BertTokenizer
import torch

# Step 2: Device configuration
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Step 3: Define the path to the extracted model folder
model_path = r"D:\bert-finetuned-sentiment"  # Update this if your folder name is different

# Step 4: Load the fine-tuned model and tokenizer from the local folder
model = BertForSequenceClassification.from_pretrained(model_path)
tokenizer = BertTokenizer.from_pretrained(model_path)
model.to(device)

# Step 5: Define class labels (update as per your training)
label_map = {
    0: "negative",
    1: "positive"
    # Add more if it's a multi-class model
}

# Step 6: Sentiment prediction function
def predict_sentiment(model, tokenizer, texts, device):
    model.eval()
    encoded = tokenizer(
        texts,
        return_tensors='pt',
        padding=True,
        truncation=True,
        max_length=512
    ).to(device)

    with torch.no_grad():
        outputs = model(**encoded)
        predictions = torch.argmax(outputs.logits, dim=-1)

    return [label_map[pred.item()] for pred in predictions]

# Step 7: Example inputs
random_reviews = [
    "Jurassic World: Dominion is a loud, bloated mess that tries to do too much and ends up doing very little well. The plot is convoluted, the villain is forgettable, and the dinosaurs — the stars of the show — feel sidelined by unnecessary subplots. It relies heavily on nostalgia without earning it, and even the returning OG characters feel underutilized.That said, the visuals are still top-notch, and there are a handful of sequences that genuinely capture the thrill and tension fans expect. Jeff Goldblum brings some much-needed charisma, and a few action set pieces are fun in a popcorn-movie kind of way. But overall, it’s hard not to feel disappointed. It’s not terrible, just a major missed opportunity"
]

# Step 8: Get predictions
results = predict_sentiment(model, tokenizer, random_reviews, device)

# Step 9: Show results
for review, sentiment in zip(random_reviews, results):
    print(f"Review: \"{review}\" => Sentiment: {sentiment}")
```

```python

from transformers import BertForSequenceClassification, BertTokenizer
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_path = r"D:\bert-finetuned-sentiment"  
model = BertForSequenceClassification.from_pretrained(model_path)
tokenizer = BertTokenizer.from_pretrained(model_path)
model.to(device)
label_map = {
    0: "negative",
    1: "positive"
}

def predict_sentiment(model, tokenizer, texts, device):
    model.eval()
    encoded = tokenizer(
        texts,
        return_tensors='pt',
        padding=True,
        truncation=True,
        max_length=512
    ).to(device)

    with torch.no_grad():
        outputs = model(**encoded)
        predictions = torch.argmax(outputs.logits, dim=-1)

    return [label_map[pred.item()] for pred in predictions]

print("Enter a movie review (type 'exit' to quit):")
while True:
    user_input = input(">>> ")
    if user_input.lower() == "exit":
        
        break
    sentiment_result = predict_sentiment(model, tokenizer, [user_input], device)
    print(f"Sentiment: {sentiment_result[0]}\n")
```
